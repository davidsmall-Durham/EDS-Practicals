###**Introducing the NetCDF file format**

**NetCDF (Network Common Data Form)** is a file format that is ideal for storing multidimensional (i.e. 3D or 4D) scientific data, especially environmental data. This can include data variables such as temperature, precipitation, salinity, wind speed, or near enough any environmental property you can think of. One of the main advantages of NetCDF is that it is self-describing, that is a NetCDF file includes information about the data it contains. The four key aspects of this are:

Component|Category|Description|Key components
---------|---------|----------|--------------
**Dimensions**|N/A|Describes the spatial and temporal extent of the data arrays.|Name and Length (e.g., Latitude, Longitude, Time, Depth/Elevation).
**Variables**|**Coordinate Variables**|Holds the values for each axis (dimension).|Name, Dimension, and Values (e.g., Latitude variable holding values from 90°N to 90°S).
 -|**Data Variables**|Holds the actual environmental or scientific measurements.|Fits the global dimensions (e.g., A global array for 20 years of annual temperature data).
 **Attributes**|N/A|Provides specific metadata for each individual variable.|Name and Value (usually text). Examples: units (Celsius, metres), valid_min/valid_max, and creator comments.
 **Global Attributes**|N\A|Provides high-level metadata applicable to the entire file.|Name and Value (usually text). Examples: Dataset name, version number, references, and naming conventions (e.g., CF Conventions).

```python
#@title ##### This cell creates our plot (ignore the download warnings and be patient)

# Set up a template for the matplotlib (plt) figure and set it's size (15 inches wide, 8 inches high)
plt.figure(figsize=(15, 8))

# Set the map projection for the axes
ax = plt.axes(projection=ccrs.EqualEarth())

# Plot the data
plot = ds.sst_celsius.plot(                                         # here we tell python what we want to plot (take our dataset (ds) called sst_celsius and plot it with the parameters we are going to define in parentheses)
    ax=ax,                                                          # Tells python what axes to plot in
    transform=ccrs.PlateCarree(),                                   # We almost ALWAYS use transform=ccrs.PlateCarree() because the NetCDF coordinates are Lat/Lon pairs.
    cmap='RdYlBu_r',                                                # This is our colour scheme (or colourmap)
    cbar_kwargs={'label': 'ADD A LABEL HERE (°C)', 'shrink': 0.8},  # This controls the label of the scale bar, add one here. Shrink makes it 80% of total plot height
)

# Add Earth details using the cartopy package (which we imported as cfeature)
ax.add_feature(cfeature.COASTLINE, linewidth=6)                          # Adds a coastline, it is too thick at the moment!!
ax.add_feature(cfeature.LAND, facecolor='red')                           # Make the land a better colour, "lightgrey" perhaps.
ax.add_feature(cfeature.BORDERS, linestyle=':')                          # Adds international borders
ax.gridlines(draw_labels=True, dms=True, x_inline=False, y_inline=False) # Controls gridlines and their labels (dms = degrees minutes seconds, could also use decimal degrees by omitting this option)

plt.title(f"ADD YOUR TITLE HERE", fontsize=2)                            # Add a title for our plot and make it more prominent
plt.show()                                                               # We have to tell Python to show the plot
```
